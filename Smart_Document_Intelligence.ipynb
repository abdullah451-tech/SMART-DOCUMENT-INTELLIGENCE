```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

def semantic_search(question, top_k=5):

    # Convert the question into an embedding
    question_embedding = model.encode([question])

    # Compare the question with all document embeddings
    similarities = cosine_similarity(question_embedding, embeddings)[0]

    # Find the indexes of the most relevant chunks
    best_indexes = np.argsort(similarities)[::-1][:top_k]

    print("Question:", question)
    print("\nMost relevant document sections:\n")

    for index in best_indexes:
        print("Similarity score:", round(float(similarities[index]), 3))
        print(chunks[index][:700])
        print("-" * 60)
```

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

def retrieve_documents(question, top_k=5):

    # Convert the question into an embedding
    question_embedding = model.encode([question])

    # Compare the question with saved document embeddings
    similarities = cosine_similarity(
        question_embedding,
        loaded_embeddings
    )[0]

    # Find the top matching document sections
    best_indices = np.argsort(similarities)[::-1][:top_k]

    results = []

    for index in best_indices:
        results.append({
            "chunk": loaded_chunks[index],
            "similarity_score": round(float(similarities[index]), 3)
        })

    return results

print("Document retrieval function created successfully!")
```

```python
# Test document retrieval
question = "What are the major agricultural reforms in India?"

results = retrieve_documents(question, top_k=5)

print("Question:", question)
print("\nRetrieved document sections:\n")

for number, result in enumerate(results, start=1):
    print(f"--- Result {number} ---")
    print("Similarity score:", result["similarity_score"])
    print(result["chunk"][:700])
    print("-" * 60)
```

```python
# Combine retrieved sections into one context

def create_context(results):

    context = ""                                                                            #Why we're doing this: The AI model needs the relevant document sections as context before it can generate an answer. This is the bridge between retrieval and generation in RAG.

    for number, result in enumerate(results, start=1):
        context += f"\nSource {number}:\n"
        context += result["chunk"]
        context += "\n"

    return context


context = create_context(results)

print("Context created successfully!")
print("Context length:", len(context), "characters")
print("\nFirst 1000 characters:\n")
print(context[:1000])
```

```python
import re

def generate_answer(question, results):
    question_words = set(re.findall(r'\b[a-zA-Z]{3,}\b', question.lower()))

    scored_sentences = []

    for result in results:
        sentences = re.split(r'(?<=[.!?])\s+', result["chunk"])

        for sentence in sentences:
            words = set(re.findall(r'\b[a-zA-Z]{3,}\b', sentence.lower()))
            score = len(question_words.intersection(words))

            if score > 0:
                scored_sentences.append((score, sentence.strip()))

    scored_sentences.sort(key=lambda item: item[0], reverse=True)

    best_sentences = []
    for score, sentence in scored_sentences:
        if sentence and sentence not in best_sentences:
            best_sentences.append(sentence)

        if len(best_sentences) == 3:
            break

    if not best_sentences:
        return "I couldn't find a relevant answer in the retrieved document sections."

    return " ".join(best_sentences)
```

```python
%%writefile search_engine.py

import json
import numpy as np
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# Load saved document data
with open("outputs/chunks.json", "r", encoding="utf-8") as file:
    chunks = json.load(file)

# Load saved embeddings
embeddings = np.load("outputs/document_embeddings.npy")

# Load embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")

def search_document(question):

    question_embedding = model.encode([question])

    similarities = cosine_similarity(
        question_embedding,
        embeddings
    )[0]

    top_indices = similarities.argsort()[-3:][::-1]

    results = []

    for index in top_indices:
        results.append({
            "page": chunks[index]["page"],
            "text": chunks[index]["text"],
            "similarity": float(similarities[index])
        })

    return results
```

```python
import fitz
import os
import json
import hashlib

documents_folder = "documents"
all_chunks = []
processed_files = set()

chunk_size = 1000
overlap = 200

for filename in os.listdir(documents_folder):

    if not filename.lower().endswith(".pdf"):
        continue

    pdf_path = os.path.join(documents_folder, filename)

    # Identify duplicate PDF files
    with open(pdf_path, "rb") as file:
        file_hash = hashlib.sha256(file.read()).hexdigest()

    if file_hash in processed_files:
        print("Skipped duplicate:", filename)
        continue

    processed_files.add(file_hash)

    print("Processing:", filename)

    pdf = fitz.open(pdf_path)

    for page_number, page in enumerate(pdf, start=1):

        page_text = page.get_text().strip()

        start = 0

        while start < len(page_text):

            end = start + chunk_size
            chunk_text = page_text[start:end]

            if chunk_text.strip():

                all_chunks.append({
                    "source": filename,
                    "page": page_number,
                    "text": chunk_text
                })

            start += chunk_size - overlap

    pdf.close()

# Save the new multi-document chunks
output_path = "outputs/all_documents_chunks.json"

with open(output_path, "w", encoding="utf-8") as file:
    json.dump(all_chunks, file, ensure_ascii=False, indent=2)

print("\nMULTI-DOCUMENT PROCESSING COMPLETE")
print("Total chunks:", len(all_chunks))
print("Unique PDFs processed:", len(processed_files))
print("Saved to:", output_path)

print("\nPDF sources included:")

for source in sorted(set(chunk["source"] for chunk in all_chunks)):
    print("-", source)
```
